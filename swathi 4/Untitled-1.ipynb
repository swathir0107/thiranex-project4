```python
df = df.drop_duplicates()

print("Dataset shape after removing duplicates:", df.shape)
```

```python
df.dtypes
```

```python
df['Date'] = pd.to_datetime(df['Date'])
```

```python
df['Date'].dtype
```

```python
df['Year'] = df['Date'].dt.year
```

```python
df['Year']
```

```python
df['Month'] = df['Date'].dt.month
```

```python
df['Month']
```

```python
df['Month Name'] = df['Date'].dt.month_name()
```

```python
df['Month Name']
```

```python
df['Day'] = df['Date'].dt.day
```

```python
df['Day']
```

```python
df['Day of Week'] = df['Date'].dt.day_name()
```

```python
df['Day of Week']
```

```python
df.head()
```

```python
total_sales = df['Total Amount'].sum()

print("Total Sales:", total_sales)
```

```python
average_sales = df['Total Amount'].mean()

print("Average Sales:", round(average_sales, 2))
```

```python
maximum_sales = df['Total Amount'].max()

print("Maximum Transaction Amount:", maximum_sales)
```

```python
minimum_sales = df['Total Amount'].min()

print("Minimum Transaction Amount:", minimum_sales)
```

```python
total_quantity = df['Quantity'].sum()

print("Total Quantity Sold:", total_quantity)
```

```python
unique_customers = df['Customer ID'].nunique()

print("Number of Unique Customers:", unique_customers)
```

```python
gender_sales = df.groupby('Gender')['Total Amount'].sum()

gender_sales
```

```python
plt.figure(figsize=(8,5))

sns.barplot(
    x=gender_sales.index,
    y=gender_sales.values
)

plt.title("Total Sales by Gender")
plt.xlabel("Gender")
plt.ylabel("Total Amount")
plt.show()
```

```python
gender_transactions = df['Gender'].value_counts()

gender_transactions
```

```python
plt.figure(figsize=(7,5))

sns.barplot(
    x=gender_transactions.index,
    y=gender_transactions.values
)

plt.title("Number of Transactions by Gender")
plt.xlabel("Gender")
plt.ylabel("Number of Transactions")
plt.show()
```

```python
category_sales = (
    df.groupby('Product Category')['Total Amount']
    .sum()
    .sort_values(ascending=False)
)

category_sales
```

```python
plt.figure(figsize=(8,5))

sns.barplot(
    x=category_sales.index,
    y=category_sales.values
)

plt.title("Sales by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Total Sales")
plt.show()
```

```python
category_quantity = (
    df.groupby('Product Category')['Quantity']
    .sum()
    .sort_values(ascending=False)
)

category_quantity
```

```python
plt.figure(figsize=(8,5))

sns.barplot(
    x=category_quantity.index,
    y=category_quantity.values
)

plt.title("Quantity Sold by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Quantity")
plt.show()
```

```python
average_price_category = (
    df.groupby('Product Category')['Price per Unit']
    .mean()
)

average_price_category
```

```python
plt.figure(figsize=(8,5))

sns.barplot(
    x=average_price_category.index,
    y=average_price_category.values
)

plt.title("Average Price per Unit by Category")
plt.xlabel("Product Category")
plt.ylabel("Average Price")
plt.show()
```

```python
monthly_sales = (
    df.groupby(['Year', 'Month'])['Total Amount']
    .sum()
    .reset_index()
)

monthly_sales
```

```python
monthly_sales['Date'] = pd.to_datetime(
    monthly_sales['Year'].astype(str) + '-' +
    monthly_sales['Month'].astype(str) + '-01'
)

monthly_sales
```

```python
plt.figure(figsize=(14,6))

sns.lineplot(
    data=monthly_sales,
    x='Date',
    y='Total Amount',
    marker='o'
)

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.show()
```

```python
day_sales = (
    df.groupby('Day of Week')['Total Amount']
    .sum()
)

day_sales
```

```python
plt.figure(figsize=(10,5))

sns.barplot(
    x=day_sales.index,
    y=day_sales.values
)

plt.title("Sales by Day of Week")
plt.xlabel("Day")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(
    df['Price per Unit'],
    bins=10,
    kde=True
)

plt.title("Price per Unit Distribution")
plt.xlabel("Price per Unit")
plt.ylabel("Frequency")
plt.show()
```

```python
best_category = category_sales.idxmax()

print("Best Performing Category:", best_category)
```

```python
best_month = monthly_sales.loc[
    monthly_sales['Total Amount'].idxmax()
]

print("Best Sales Month:", best_month['Month'])
print("Sales:", best_month['Total Amount'])
```

```python
customer_sales = (
    df.groupby('Customer ID')['Total Amount']
    .sum()
    .sort_values(ascending=False)
)

print("Highest Spending Customer:")
print(customer_sales.head(10))
```

```python
top_customers = customer_sales.head(10)

plt.figure(figsize=(10,6))

sns.barplot(
    x=top_customers.values,
    y=top_customers.index
)

plt.title("Top 10 Customers by Total Spending")
plt.xlabel("Total Spending")
plt.ylabel("Customer ID")
plt.show()
```

```python
print("========== RETAIL SALES INSIGHTS ==========")

print("Total Sales:", df['Total Amount'].sum())

print("Average Transaction Amount:",
      round(df['Total Amount'].mean(), 2))

print("Total Quantity Sold:",
      df['Quantity'].sum())

print("Number of Unique Customers:",
      df['Customer ID'].nunique())

print("Best Product Category:",
      category_sales.idxmax())

print("Highest Spending Customer:",
      customer_sales.idxmax())

print("Best Sales Month:",
      monthly_sales.loc[
          monthly_sales['Total Amount'].idxmax(),
          'Month'
      ])

print("Highest Transaction Amount:",
      df['Total Amount'].max())
```
